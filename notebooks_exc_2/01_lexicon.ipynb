## 3. Save resources to JSON


```python
import json
from pathlib import Path

resources = {
    "language": "English",
    "notes": "Simple SVO with regular verbs only.",
    "nouns": NOUNS,
    "adjectives": ADJECTIVES,
    "verbs": VERBS,
}

output_path = Path.cwd() / "svo_resources.json"
output_path.parent.mkdir(parents=True, exist_ok=True)
output_path.write_text(json.dumps(resources, indent=2, ensure_ascii=True))

print(f"Saved: {output_path}")
```
