# Swadesh List Comparison

The [Swadesh list](https://en.wikipedia.org/wiki/Swadesh_list) is a classic tool in linguistics used to compare the basic vocabulary of different languages. It contains a set of universal concepts that are likely to be present in any language.

In this notebook, we will:
1.  Identify the top 50 most frequent nouns from our German corpus.
2.  Compare this list with the German Swadesh list.
3.  Analyze the overlap to see how much of the most common vocabulary is covered by this foundational word list.

```python
import os
import pandas as pd
import spacy
from tqdm import tqdm

# --- Configuration ---
DATA_DIR = "../data"
WORD_COUNTS_FILE = "word_counts_with_stopwords.csv"
TOP_N_NOUNS = 50

# --- Load Data ---
word_counts_path = os.path.join(DATA_DIR, WORD_COUNTS_FILE)
df_words = pd.read_csv(word_counts_path)

# --- German Swadesh List ---
# A version of the Swadesh list for German
swadesh_list = [
    "ich", "du", "er", "wir", "ihr", "sie", "dieser", "jener", "hier", "dort",
    "wer", "was", "wo", "wann", "wie", "nicht", "alle", "viele", "einige", "wenige",
    "andere", "eins", "zwei", "drei", "vier", "fünf", "groß", "lang", "breit", "dick",
    "schwer", "klein", "kurz", "eng", "dünn", "frau", "mann", "mensch", "kind", "frau",
    "ehemann", "mutter", "vater", "tier", "fisch", "vogel", "hund", "laus", "schlange",
    "wurm", "baum", "wald", "stock", "frucht", "samen", "blatt", "wurzel", "rinde",
    "blume", "gras", "seil", "haut", "fleisch", "blut", "knochen", "fett", "ei",
    "horn", "schwanz", "feder", "haar", "kopf", "ohr", "auge", "nase", "mund",
    "zahn", "zunge", "fingernagel", "fuß", "bein", "knie", "hand", "flügel", "bauch",
    "gedärme", "hals", "rücken", "brust", "herz", "leber", "trinken", "essen", "beißen",
    "saugen", "spucken", "erbrechen", "blasen", "atmen", "lachen", "sehen", "hören",
    "wissen", "denken", "riechen", "fürchten", "schlafen", "leben", "sterben", "töten",
    "kämpfen", "jagen", "schlagen", "schneiden", "teilen", "stechen", "kratzen", "graben",
    "schwimmen", "fliegen", "gehen", "kommen", "liegen", "sitzen", "stehen", "geben",
    "halten", "drücken", "reiben", "waschen", "wischen", "ziehen", "werfen", "binden",
    "nähen", "zählen", "sagen", "singen", "spielen", "schwimmen", "fließen", "frieren",
    "schwellen", "sonne", "mond", "stern", "wasser", "regen", "fluss", "see", "meer",
    "salz", "stein", "sand", "staub", "erde", "wolke", "nebel", "himmel", "wind",
    "schnee", "eis", "rauch", "feuer", "asche", "brennen", "weg", "berg", "rot",
    "grün", "gelb", "weiß", "schwarz", "nacht", "tag", "jahr", "warm", "kalt",
    "voll", "neu", "alt", "gut", "schlecht", "faul", "gerade", "rund", "scharf",
    "stumpf", "glatt", "nass", "trocken", "richtig", "nah", "fern", "rechts", "links",
    "bei", "in", "mit", "und", "wenn", "weil", "name"
]
swadesh_set = set(swadesh_list)

# --- Load spaCy model ---
nlp = spacy.load("de_core_news_sm")

# --- Find Top Nouns ---
nouns = []
# Use tqdm to show progress as this can take a moment
for word in tqdm(df_words['word'], desc="Finding nouns"):
    # We create a doc for each word to check its POS tag
    doc = nlp(str(word))
    if len(doc) > 0 and doc[0].pos_ == 'NOUN':
        nouns.append(word)
    
    if len(nouns) >= TOP_N_NOUNS:
        break

top_nouns_set = set(nouns)

# --- Compare Lists ---
common_words = top_nouns_set.intersection(swadesh_set)
swadesh_nouns_in_top = [word for word in swadesh_list if word in top_nouns_set]
swadesh_nouns_not_in_top = [word for word in swadesh_list if nlp(word)[0].pos_ == 'NOUN' and word not in top_nouns_set]


# --- Display Results ---
print(f"--- Top {TOP_N_NOUNS} Most Frequent Nouns ---")
print(", ".join(sorted(list(top_nouns_set))))

print(f"\n--- Nouns in both Top {TOP_N_NOUNS} and Swadesh List ---")
if common_words:
    print(", ".join(sorted(list(common_words))))
else:
    print("None")

print(f"\n--- Swadesh List Nouns NOT in the Top {TOP_N_NOUNS} ---")
print(", ".join(sorted(swadesh_nouns_not_in_top)))

# --- Calculate Overlap ---
overlap_percentage = (len(common_words) / len(top_nouns_set)) * 100 if top_nouns_set else 0
print(f"\nOverlap: {len(common_words)} of the top {TOP_N_NOUNS} nouns are in the Swadesh list ({overlap_percentage:.2f}%).")
```
